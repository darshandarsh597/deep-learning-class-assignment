```python
import numpy as np
```

```python
X = np.random.rand(1, 3, 5, 5)
```

```python
num_filters = 2
filter_size = 3
stride = 1
padding = 1

# Random Filters
W = np.random.randn(num_filters, 3, filter_size, filter_size)

# Bias
b = np.zeros((num_filters, 1))
```

```python
def im2col(input_data, filter_h, filter_w, stride=1, pad=0):

    N, C, H, W = input_data.shape

    out_h = (H + 2 * pad - filter_h) // stride + 1
    out_w = (W + 2 * pad - filter_w) // stride + 1

    img = np.pad(input_data,
                 [(0,0), (0,0), (pad,pad), (pad,pad)],
                 mode='constant')

    cols = np.zeros((N, C, filter_h, filter_w, out_h, out_w))

    for y in range(filter_h):
        y_max = y + stride * out_h

        for x in range(filter_w):
            x_max = x + stride * out_w

            cols[:, :, y, x, :, :] = img[:, :, y:y_max:stride, x:x_max:stride]

    cols = cols.transpose(0,4,5,1,2,3).reshape(N*out_h*out_w, -1)

    return cols
```

```python
def im2col(input_data, filter_h, filter_w, stride=1, pad=0):

    N, C, H, W = input_data.shape

    out_h = (H + 2 * pad - filter_h) // stride + 1
    out_w = (W + 2 * pad - filter_w) // stride + 1

    img = np.pad(input_data,
                 [(0,0), (0,0), (pad,pad), (pad,pad)],
                 mode='constant')

    cols = np.zeros((N, C, filter_h, filter_w, out_h, out_w))

    for y in range(filter_h):
        y_max = y + stride * out_h

        for x in range(filter_w):
            x_max = x + stride * out_w

            cols[:, :, y, x, :, :] = img[:, :, y:y_max:stride, x:x_max:stride]

    cols = cols.transpose(0,4,5,1,2,3).reshape(N*out_h*out_w, -1)

    return cols
```

```python
def conv_forward(X, W, b, stride, pad):
    N, C, H, W_in = X.shape
    F, C, HH, WW = W.shape

    out_h = (H + 2 * pad - HH) // stride + 1
    out_w = (W_in + 2 * pad - WW) // stride + 1

    col = im2col(X, HH, WW, stride, pad)
    col_W = W.reshape(F, -1).T

    out = np.dot(col, col_W) + b.T
    out = out.reshape(N, out_h, out_w, F).transpose(0, 3, 1, 2)
    return out

output = conv_forward(X, W, b, stride, padding)

print("Input Shape :", X.shape)
print("Filter Shape:", W.shape)
print("Output Shape:", output.shape)

print("\nOutput Feature Maps:")
print(output)
```
