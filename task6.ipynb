Bidirectional LSTM from Scratch

```python
import torch
import numpy as np

torch.manual_seed(42)
```

Input Sequence
 (Sequence Length, Input Features)

```python
seq_len = 5
input_size = 3
hidden_size = 4

X = torch.randn(seq_len, input_size)
```

ctivation Functions

```python
def sigmoid(x):
    return 1 / (1 + torch.exp(-x))

def tanh(x):
    return torch.tanh(x)
```

Initialize Weights

```python
def initialize():

    W_x = torch.randn(input_size, hidden_size)
    W_h = torch.randn(hidden_size, hidden_size)
    b = torch.zeros(hidden_size)

    return W_x, W_h, b

# Input Gate
Wxi, Whi, bi = initialize()

# Forget Gate
Wxf, Whf, bf = initialize()

# Output Gate
Wxo, Who, bo = initialize()

# Cell Candidate
Wxc, Whc, bc = initialize()
```

```python
def lstm_forward(sequence):

    h = torch.zeros(hidden_size)
    c = torch.zeros(hidden_size)

    outputs = []

    for t in range(sequence.shape[0]):

        x = sequence[t]

        i = sigmoid(torch.matmul(x, Wxi) + torch.matmul(h, Whi) + bi)

        f = sigmoid(torch.matmul(x, Wxf) + torch.matmul(h, Whf) + bf)

        o = sigmoid(torch.matmul(x, Wxo) + torch.matmul(h, Who) + bo)

        g = tanh(torch.matmul(x, Wxc) + torch.matmul(h, Whc) + bc)

        c = f * c + i * g

        h = o * tanh(c)

        outputs.append(h)

    return torch.stack(outputs)
```

```python
reverse_sequence = torch.flip(X, dims=[0])


forward_output = lstm_forward(X)

backward_output = lstm_forward(reverse_sequence)

backward_output = torch.flip(backward_output, dims=[0])
```

```python
BiLSTM_output = torch.cat(
    (forward_output, backward_output),
    dim=1
)
```

 Display Results

```python
print("Input Sequence Shape")
print(X.shape)

print("\nForward Output Shape")
print(forward_output.shape)

print("\nBackward Output Shape")
print(backward_output.shape)

print("\nBiLSTM Output Shape")
print(BiLSTM_output.shape)

print("\nForward Hidden States")
print(forward_output)

print("\nBackward Hidden States")
print(backward_output)

print("\nFinal BiLSTM Output")
print(BiLSTM_output)
```
